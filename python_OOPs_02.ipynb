```python
class student:
  """This is Python Object-Oriented Programming
     --> we only defind variables and methodes"""# -->documentation
  def __init__(self): #methode 1 -- instance methode or called as constructor
    print("self",id(self))
    self.name="harsh" # -variables
    self.rollno=101  # -variables
    self.marks=98   # -variables

  def talk(self): # methode 2
    print(f"Hello. I am {self.name}")
    print(f"my roll number is {self.rollno}")
    print(f"My marks are {self.marks}")
```

```python
s1=student()
#reference variable = call name()--> for creating a object
```

```python
#access the class varaiables and methodes
print(s1.marks)
print(s1.rollno)
print(s1.name)
s1.talk()
```

```python
# self and s1
print("s1",id(s1))
```

```python
# pass the parameter to object
class student1:
  """This is Python Object-Oriented Programming
     -->In this me create one class and mulitple refernce
        variable """# -->documentation
  def __init__(self,name,rollno,marks):
    self.name=name
    self.rollno=rollno
    self.marks=marks

  def talk(self): # methode 2
    print(f"Hello. I am {self.name}")
    print(f"my roll number is {self.rollno}")
    print(f"My marks are {self.marks}")

s1=student1("Harsh",19,87)# pass the parameter to object
```

```python
print(s1.marks)
print(s1.name)
print(s1.rollno)
s1.talk()
```

```python
s2=student1("satyajita",18,80)
```

```python
print(s2.marks)
print(s2.name)
print(s2.rollno)
s1.talk()
```

```python
class Employees:
  def __init__(self,name,id,salary):
    self.name = name
    self.id = id
    self.salary = salary

  def info_about_employees(self):
    print("employe id \t   = \t",self.id)
    print("Name of employee   = \t",self.name)
    print("Salary of employee = \t",self.salary)
```

```python
E1=Employees(name='Harsh',id=101,salary=198000)
E1.salary
E1.name
E1.info_about_employees()
```

```python
E2=Employees(name='shrenivas',id=102,salary=180000)
E2.salary
E2.name
E2.info_about_employees()
```
